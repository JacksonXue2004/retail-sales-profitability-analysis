```python
def remove_consecutive_duplicates(x):
    return [i for i,j in zip(x,x[1:] + [None]) if i != j]
```

```python
segment_transition_clean = segment_transition.apply(
    remove_consecutive_duplicates
)
```

```python
def create_transitions(segments):
    return [
        f"{segments[i]} → {segments[i+1]}"
        for i in range(len(segments) - 1)
    ]
```

```python
customer_transitions = segment_transition_clean.apply(
    create_transitions
)
```

```python
discount_analysis = df_clean.copy()

discount_analysis["Discount_Band"] = pd.cut(
    discount_analysis["Discount_Percentage"],
    bins=[-0.01, 0, 5, 10, 15, 20, 1000],
    labels=[
        "0%",
        "1-5%",
        "6-10%",
        "11-15%",
        "16-20%",
        "20%+"
    ]
)

discount_analysis = (
    discount_analysis
    #observed=False：关键参数！因为 Discount_Band 是分类类型（Categorical），
    #设置 observed=False 可以确保即使某个折扣区间（例如 16-20%）在数据集中没有任何订单，
    #该区间也会呈现在结果表中（对应的数据显示为 0 或 NaN），保证分组标签的完整性。
    .groupby("Discount_Band", observed=False)
    .agg(
        Sales=("Sales_Amount", "sum"),
        Profit=("Profit", "sum"),
        Orders=("Order_ID", "nunique"),
        Avg_Discount=("Discount_Percentage", "mean")
    )
    .reset_index()
)

discount_analysis["Profit_Margin"] = (
    discount_analysis["Profit"]
    / discount_analysis["Sales"]
    * 100
)

discount_analysis["AOV"] = (
    discount_analysis["Sales"]
    / discount_analysis["Orders"]
)

discount_analysis
```

```python
# 创建 Discount_Band
df_clean["Discount_Band"] = pd.cut(
    df_clean["Discount_Percentage"],
    bins=[-0.01, 0, 5, 10, 15, 20, 1000],
    labels=["0%", "1-5%", "6-10%", "11-15%", "16-20%", "20%+"]
)

# Discount × Product Category
discount_category = (
    df_clean
    .groupby(["Product_Category", "Discount_Band"], observed=False)
    .agg(
        Sales=("Sales_Amount", "sum"),
        Profit=("Profit", "sum"),
        Orders=("Order_ID", "nunique")
    )
    .reset_index()
)

# 利润率
discount_category["Profit_Margin"] = (
    discount_category["Profit"]
    / discount_category["Sales"]
    * 100
)

# 每个 Category 内部的销售占比
discount_category["Sales_Share"] = (
    discount_category["Sales"]
    / discount_category.groupby("Product_Category")["Sales"].transform("sum")
    * 100
)

discount_category
```
